**Проблема:** Мы видим, что 80% людей уходят после просмотра товара.

**Гипотеза:** Если мы добавим на карточку товара блок «С этим часто покупают» (Cross-sell), это увеличит конверсию в добавление в корзину (add_to_cart) на 1% (с 20.7% до 21.7%).

**Метрика:** CR из view_item в add_to_cart.

**Группы:** Control (старая карточка) и Test (новая карточка с рекомендациями).

```python
import numpy as np
from statsmodels.stats.proportion import proportions_ztest
```

```python
# Количество корзин: [Control, Test]
successes = [6288, 6251]

# Количество просмотров: [Control, Test]
nobs = [30688, 30564]

# Считаем Z-тест для пропорций
stat, pval = proportions_ztest(successes, nobs)

print(f"Z-статистика: {stat:.4f}")
print(f"p-value: {pval:.4f}")

alpha = 0.05 # Уровень значимости

if pval < alpha:
    print("Разница статистически значима. Мы отвергаем нулевую гипотезу. Фичу можно катить в прод!")
else:
    print("Разница НЕ является статистически значимой. Мы не можем отвергнуть нулевую гипотезу. Фича не работает.")
```
